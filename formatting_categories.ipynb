```python
import pandas as pd
```

```python
# Read the datasets
df_train = pd.read_csv('Dataset/train_cleaned.csv')
df_test = pd.read_csv('Dataset/test_cleaned.csv')
df_validation = pd.read_csv('Dataset/validation_cleaned.csv')
```

```python
categories = {
    "Politics": 0,
    "Government": 0,
    "Left": 0,
    "Middle-east": 0,
    "Us": 0,
    "News": 0
}
```

```python
def get_values(categories, df):
    category_keys = [c.lower() for c in categories]
    unmatched_topics = []

    for topic_str in df['topic']:
        topic_str_l = topic_str.lower()
        matched = False
        for cat in category_keys:
            if cat in topic_str_l:
                categories[cat.capitalize()] += 1
                matched = True
                break
        if not matched:
            unmatched_topics.append(topic_str)

    if len(unmatched_topics) > 0:
        print("Unmatched Topics:")
        for t in set(unmatched_topics):
            print("-", t)

    print("Matched Topics:")
    for c in categories:
        print(c, categories[c])
```

```python
print("TRAIN")
get_values(categories, df_train)

print("\nTEST")
get_values(categories, df_test)

print("\nVALIDATION")
get_values(categories, df_validation)
```

```python
categories = {
    "Politics": 0,
    "Government": 0,
    "Left": 0,
    "Middle-east": 0,
    "Us": 0,
    "News": 0
}
```

```python
def get_values(categories, df):
    category_keys = list(categories.keys())
    unmatched_topics = []

    for key in category_keys:
        categories[key] = 0

    new_labels = []

    for topic_str in df['topic']:
        topic_str_l = topic_str.lower()
        matched = False
        for cat in category_keys:
            if cat.lower() in topic_str_l:
                categories[cat] += 1
                new_labels.append(cat)
                matched = True
                break
        if not matched:
            unmatched_topics.append(topic_str)
            new_labels.append("Unmatched")

    df['topic'] = new_labels

    print("Topics riconosciuti:")
    for c in categories:
        print(c, categories[c])
```

```python
print("TRAIN")
get_values(categories, df_train)

print("\nTEST")
get_values(categories, df_test)

print("\nVALIDATION")
get_values(categories, df_validation)
```

```python
df_train.to_csv('Dataset/train_cleaned.csv', index=False)
df_test.to_csv('Dataset/test_cleaned.csv', index=False)
df_validation.to_csv('Dataset/validation_cleaned.csv', index=False)
```
